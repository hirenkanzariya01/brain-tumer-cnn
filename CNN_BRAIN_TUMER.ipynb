```python
import os
from PIL import Image
from sklearn import utils
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
Training_data_path = '/content/drive/MyDrive/archive/Training'
Testing_data_path = '/content/drive/MyDrive/archive/Testing'
```

```python
training_label = []
training_data = []
for i in os.listdir(Training_data_path):
  for j in os.listdir(Training_data_path+'/'+i):
    training_label.append(i)
    training_data.append('/content/drive/MyDrive/archive/Training/'+i + '/' +j)

training_data , training_label = utils.shuffle(training_data, training_label)
```

```python
testing_data = []
testing_label = []
for i in os.listdir(Testing_data_path):
  for j in os.listdir(Testing_data_path+'/'+i):
    testing_data.append('/content/drive/MyDrive/archive/Testing/'+i + '/' +j)
    testing_label.append(i)


testing_data, testing_label = utils.shuffle(testing_data, testing_label)
```

```python
from tensorflow.keras.utils import img_to_array
```

```python
training_image_data = []
```

```python
for i  in training_data:
  img = Image.open(i)
  img = img.resize((224,224))
  img = img_to_array(img)
  training_image_data.append(img)
```

```python
from tensorflow.keras import layers , Sequential
from tensorflow.keras import models
from tensorflow.keras.layers import Conv2D , MaxPooling2D , Flatten , Dense , Dropout
```

```python
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(224,224,3)),
    MaxPooling2D((2,2)),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D((2,2)),
    Conv2D(128, (3,3), activation='relu'),
    MaxPooling2D((2,2)),
    Conv2D(128, (3,3), activation='relu'),
    MaxPooling2D((2,2)),
    Flatten(),
    Dense(512, activation='relu'),
    Dropout(0.5),
    Dense(256, activation='relu'),
    Dropout(0.5),
    Dense(2, activation='softmax')
])
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(np.array(training_image_data), np.array(training_label), epochs=10, batch_size=32)
```
